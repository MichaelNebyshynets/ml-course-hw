ЧАСТЬ B

```python
print("Стратегия dropna(удаление строк с пропусками)")
print(f"Размер датасета до удаления: {df.shape}")
df_dropped = df.dropna()
print(f"Размер датасета после удаления: {df_dropped.shape}")
print(f"Удалено строк: {df.shape[0]-df_dropped.shape[0]}")
```

```python
print("Стратегия fillna(заполнение пропусков)")
#создаем копию для заполнения пропусков
df_filled = df.copy()

numeric_cols = df_filled.select_dtypes(include=[np.number]).columns

for col in numeric_cols:
    if df_filled[col].isnull().any():  # проверяем, есть ли пропуски в колонке
        # ОБОСНОВАНИЕ: Используем среднее значение, так как распределение близко к нормальному
        # и нет сильных выбросов, которые могли бы исказить среднее
        mean_val = df_filled[col].mean()
        df_filled[col].fillna(mean_val, inplace=True)
        print(f"Колонка '{col}': заполнена СРЕДНИМ значением {mean_val:.2f}")
        print(f"  Причина: распределение симметричное, нет сильных выбросов")       

# Заполнение категориальных колонок
print("\nЗаполнение категориальных колонок:")
cat_cols = df.select_dtypes(include=['object']).columns

for col in cat_cols:
    if df[col].isnull().any():
        # ОБОСНОВАНИЕ: Для категориальных признаков лучше всего использовать моду,
        # так как это наиболее часто встречающееся значение
        mode_val = df[col].mode()[0]  # mode()[0] берет первое значение моды
        df[col].fillna(mode_val, inplace=True)
        print(f"Колонка '{col}': заполнена МОДОЙ '{mode_val}'")
        print(f"  Причина: мода представляет наиболее типичную категорию")
```

```python
print("Как считаем моду")

for col in cat_cols:
    if col in df.columns:  # проверяем, что колонка существует в исходных данных
        print(f"\nКолонка: {col}")
        # Вычисляем частоту встречаемости каждого значения
        value_counts = df[col].value_counts()
        print("Частота значений:")
        for val, count in value_counts.head().items():  # показываем топ-5
            print(f"  '{val}': {count} раз ({count/len(df)*100:.1f}%)")
        
        # Находим моду
        mode_val = df[col].mode()[0]
        mode_count = (df[col] == mode_val).sum()
        print(f"\nМОДА: '{mode_val}' (встречается {mode_count} раз)")
```

ЧАСТЬ С

```python
#подсчет min, max, mean, median, mode для числовых колонок(только release_year)
for col in numeric_cols:
    print(f"Базовая статистика для {col}")
    print(f"Минимальное значение: {df[col].min()}")
    print(f"Максимальное значение: {df[col].max()}")
    print(f"Среднее значение: {df[col].mean()}")
    print(f"Медиана: {df[col].median()}")
    print(f"Мода: {df[col].mode()[0]}")
```

```python
#перцентили
percentiles = [0.05, 0.25, 0.5, 0.75, 0.95]

for col in numeric_cols:
    print(f"Перцентили для {col}")
    for p in percentiles:
        value = df['release_year'].quantile(p)
        print(f"{p*100}: {value}")
```

```python
#Продвинутая статистика

for col in numeric_cols:
    #varience - дисперсия показывает разброс данных вокруг среднего значения(типо ширина колокола)
    #в данном случае высокая дисперсия говорит о том, что на нетфликсе разброс даты выхода фильмов/сериалов большое,
    #т.е. много фильмов с разных времен
    varience = df[col].var()
    print(f"Varience: {varience:.2f}")

    #skewness - асимметрия показывает насколько график перекошен относительно среднего значения(форма колокола)
    # = 0 => симметрия
    # >0 => правостороняя(хвост справа, слева больше значений)
    # <0 => левосторонняя(хвост слева, справа больше значений)
    # Тут 'release_date' имеет левостороннюю асимметрию, т.е. на нетфликс постоянно добавляют новые фильмы/сериалы,
    # а старых фильмов меньше => хвост слева 
    skewness = df[col].skew()
    print(f"Skewness: {skewness:.2f}")

    #kurtosis - эксцесс показывает куда завален график(насколько острый пик, насколько тяжелые хвосты)
    # у нетфликса высокий положительный эксцесс, т.е. там большинство данных близки к среднему, но
    # часто встречаются и отклонения
    kurtosis = df[col].kurt()
    print(f"Kurtosis: {kurtosis:.2f}")
```

```python
# 1. Распределения('release_year')

plt.figure(figsize=(12, 6))
sns.histplot(df['release_year'], kde=True, color='darkred', bins=30)
plt.title('Распределение контента по годам выпуска (Release Year)')
plt.grid(axis='y', alpha=0.3)
plt.xlabel('Год выпуска')
plt.ylabel('Количество контента')
plt.show()
```

```python
# Scatter plot(зависимость длины фильмов от года выпуска)

# Извлекаем минуты для фильмов
df['duration_min'] = df.apply(
    lambda x: int(x['duration'].split(' ')[0]) if x['type'] == 'Movie' and pd.notnull(x['duration']) else None, 
    axis=1
)
# Извлекаем сезоны для сериалов
df['seasons_count'] = df.apply(
    lambda x: int(x['duration'].split(' ')[0]) if x['type'] == 'TV Show' and pd.notnull(x['duration']) else None, 
    axis=1
)

plt.figure(figsize=(12, 6))
sns.scatterplot(data=df[df['type'] == 'Movie'], x='release_year', y='duration_min', 
                alpha=0.4, color='orange', edgecolor='w')
plt.title('Изменение длительности фильмов с течением времени', fontsize=15)
plt.xlabel('Год выпуска')
plt.ylabel('Длительность (мин)')
plt.show()
```

```python
# Box plot(поиск аномалий в длительности сериалов)

plt.figure(figsize=(10, 6))
sns.boxplot(
    x='type', 
    y='seasons_count', 
    data=df[df['type'] == 'TV Show'], 
    hue='type',           
    palette='Set2', 
    legend=False          
)
plt.title('Ящик с усами: Распределение сезонов в сериалах (поиск аномалий)', fontsize=15)
plt.ylabel('Количество сезонов')
plt.show()
```

```python
# Столбчатая диаграмма(рейтинг)

plt.figure(figsize=(12, 6))
# countplot() — сама считает количество вхождений каждой категории
# order=... — сортирует столбцы от самого популярного к самому редкому
plt.figure(figsize=(12, 6))
sns.countplot(
    data=df, 
    x='rating', 
    order=df['rating'].value_counts().index, 
    hue='rating',         # Добавляем hue, равный x
    palette='viridis', 
    legend=False          # Отключаем легенду, так как подписи и так внизу
)
plt.xticks(rotation=45)
plt.show()
```

```python
# Heatmap

plt.figure(figsize=(8, 6))
# corr() — вычисляет коэффициент корреляции Пирсона между числовыми колонками
numeric_df = df[['release_year', 'duration_min', 'seasons_count']].dropna(how='all')
corr = numeric_df.corr()
# heatmap() — визуализирует матрицу корреляций цветом
# annot=True добавляет числа (коэффициенты) внутрь квадратов
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Матрица корреляций признаков')
plt.show()
```
